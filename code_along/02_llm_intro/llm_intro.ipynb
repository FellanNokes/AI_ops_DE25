## Intro to LLMs and PydanticAI


````python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv

load_dotenv()


model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-lightning-30b-a3b",
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1",
        api_key = os.environ["NVIDIA_API"]
    ),
)

model
````

````python
from IPython.display import Markdown
agent = Agent(
    model=model,
    system_prompt="You are a funny joking bot, answer shortly in a nerdy style, also add emojis"
)
prompt = "Explain what PydanticAI is and the differnece between it and Pydantic"

result = await agent.run(prompt)

Markdown(result.output)
````

## ChefBot

````python
chef_agent = Agent(
    model=model,
        instructions=("You are a professional chef that is very good at making asian food and explaining in a pedagogical way how to create yummy dishes."
                  """
                    Answer in the following format:

                    Description 1-2 sentences of what the dish is
        
                    List of ingredients in bullet points

                    Short step by step description on how to make the dish
                  """
                  """Return results as a python code so it makes a list and loops through results, Without this part ```python
soups = and the ``` in the end """
    ),
)

result = await chef_agent.run("Give me 3 asian soups in a list")

result.output
````

````python
print(result.output)
````

````python
with open("soups.md", "w") as file:
    file.write(result.output)
````

## Thinking leve

````python
from pydantic_ai import ModelSettings

storyteller = Agent(
    model, instructions="You are a children storyteller, answer max 3 sentences",
    model_settings=ModelSettings(thinking=False, temperature=0)
)

prompt = "A rabbit called bella likes to program Python ..."

story = await storyteller.run(prompt)
````

````python
print(story.output)
````

````python
story.usage
````

````python
story.all_messages()
````

````python
story2 = await storyteller.run(prompt, timeout=60)
print(story2.output)
````

````python
story2.usage
````

## Multimodal

````python
from pathlib import Path
from pydantic_ai.messages import BinaryContent

image_joker = Agent(
    model = model,
    instructions="You are a person that can analyze an image, explain the contents and make a joke about it, max 3 sentences."
)

image_data = Path("bella.png").read_bytes()
prompt = "tell me about this dude"

result = await image_joker.run(
    user_prompt=[prompt, BinaryContent(data=image_data, media_type="image/png")]
)

result.output
````
