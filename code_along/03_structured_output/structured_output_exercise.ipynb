# Structured output from text


```python
from model import nividia_model
from pydantic_ai import Agent
from pydantic import BaseModel, Field

class Person(BaseModel):
    name: str
    organistion: str = Field("company or organisation")
    subject: str = Field("the subject the person has")

class Resilience(BaseModel):
    title: str = Field("Pick suitable headline from the input, don't hallucinate")
    descripting: str = Field("3-5 sentences summary of the content")
    contributors: list[Person]
    contact: str = Field("contact person for this event")

with open("raw_text.txt", encoding="utf-8") as file:
    raw_text = file.read()


agent = Agent(
    model= nividia_model("nvidia/nemotron-3.5-super-vl-120b-a12b"),
    instructions=("You are expert in summarizing swedish content in swedish"
                 "Take the provided content and summerize")
              )

results = await agent.run(raw_text, output_type=Resilience)
```

```python
content = results.output
content
```

## Structure a program

```python
class Kurser(BaseModel):
    name: str = Field("Name of the course")
    yhpoints: int = Field("Amount of YH points that the course yields")
    weeks: int = Field("Amount of weeks that the course runs (5 points is 1 week)")
    content: str = Field("Info on what the course contains")

class Program(BaseModel):
    title: str = Field("Title of the programm")
    about: str = Field("What the education is about")
    description: str = Field("describes what the person does when they start working")
    compentense: str = Field("What compenstenses that the program yields")
    courses: list[Kurser]
    contact_info: str

agent = Agent(
    model= nividia_model("nvidia/nemotron-3.5-super-vl-120b-a12b"),
    instructions=("You are expert in summarizing a program at Yrkeshögskola in sweden"
                 "Take the provided content and summerize")
              )

with open("raw_text3.txt", encoding="utf-8") as file:
    raw_text = file.read()

results = await agent.run(raw_text, output_type=Program)
```
